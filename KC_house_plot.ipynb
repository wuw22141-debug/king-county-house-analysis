```python
df = houses.loc[houses.bedrooms.isin([2,3,4,5]),["bedrooms","price1000","year_month"]]
```

# Plot multiple line charts on the same figure

```python
monthly_by_bedroom = df.groupby(["year_month", "bedrooms"])["price1000"].median().unstack()
```

```python
monthly_by_bedroom
```

```python
for bedrooms in [2, 3, 4,5]:
    plt.plot(
        monthly_by_bedroom.index,
        monthly_by_bedroom[bedrooms],
        marker="o",
        label=f"{bedrooms} bedrooms"
    )

plt.title("Monthly Median Price by Bedrooms")
plt.xlabel("Month")
plt.ylabel("Median Price, Thousand Dollars")

plt.xticks(rotation=45)
plt.legend()
plt.grid(alpha=0.3)
```
